```python
import os
import numpy as np
from matplotlib import pyplot as plt
from matplotlib import image as mpimg
from mpl_toolkits.axes_grid1.inset_locator import zoomed_inset_axes, mark_inset
```

```python
output_path = "../../../output"

def render_image(img_path, ax):
    img = mpimg.imread(img_path)
    img = np.flipud(img)
    ax.imshow(img, origin="lower")
    ax.axis("off")

    # Inset 1
    axins1 = zoomed_inset_axes(ax, zoom=6, loc="lower left")
    axins1.imshow(img)   # <- wichtig!
    x1, x2, y1, y2 = 400, 450, 274, 324
    axins1.set_xlim(x1, x2)
    axins1.set_ylim(y1, y2)
    axins1.tick_params(left=False, bottom=False, labelleft=False, labelbottom=False)
    mark_inset(ax, axins1, loc1=4, loc2=1, fc="none", ec="red")

    # Inset 2
    axins2 = zoomed_inset_axes(ax, zoom=6, loc="upper right")
    axins2.imshow(img, origin="upper")
    x1, x2, y1, y2 = 370, 420, 804, 854
    axins2.set_xlim(x1, x2)
    axins2.set_ylim(y1, y2)
    axins2.tick_params(left=False, bottom=False, labelleft=False, labelbottom=False)
    mark_inset(ax, axins2, loc1=2, loc2=3, fc="none", ec="red")

def path(method, scene, cam_num):
    return os.path.join(output_path, f"{method}_{cam_num}_white", "results", scene, "test", "ours_30000", "renders", "00004.png")

methods = ["separate", "tandem", "fused"]
scene = "composite"
cam_num = 160

paths = [
    os.path.join(output_path, "separate_160_white", "results", "unpolarized", "test", "ours_30000", "gt", "00004.png"),
    os.path.join(output_path, "separate_160_white", "results", "unpolarized", "test", "ours_30000", "renders", "00004.png"),
    os.path.join(output_path, "separate_160_white", "results", "composite", "test", "ours_30000", "renders", "00004.png"),
    os.path.join(output_path, "tandem_160_white", "results", "composite", "test", "ours_30000", "renders", "00004.png"),
    os.path.join(output_path, "fused_160_white", "results", "composite", "test", "ours_30000", "renders", "00004.png"),
]

names = ["gt", "unpolarized", "separate", "tandem", "fused"]

for img_path, name in zip(paths, names):
    fig, ax = plt.subplots(1, 1, figsize=(5, 5))
    render_image(img_path, ax)
    fig.savefig(f"3dgs_{name}_inset.png", bbox_inches='tight', dpi=300)
    plt.show()

# fig, ax = plt.subplots(1, 3, figsize=(15, 5))
# for i, method in enumerate(methods):
#     img_path = path(method, scene, cam_num)
#     fig, ax = plt.subplots(1, 1, figsize=(5, 5))
#     render_image(img_path, ax)
#     fig.savefig(f"3dgs_{method}_inset.png", bbox_inches='tight', dpi=300)
#     plt.show()
```
